```python
from gdtcr import EmbeddingPipeline

pipe = EmbeddingPipeline(
    output_root="../output/binding_prediction",
    esmc_trd_weights   = "../output/esmc/ft32/model.pt",
    esmc_trg_weights   = "../output/esmc/ft33/model.pt",
    probert_trd_model  = "../output/probert/ft03/model.pth",
    probert_trg_model  = "../output/probert/ft04/model.pth",
    probert_base_path="../models/proteinbert",
    esmc_chunk_size=1000,
    probert_batch_size=64,
    probert_max_length=45,
)

results = pipe.run(
    tcr_csv="../output/data_processing/binding_v2/tcr_cleaned_full_pre.csv",
    out_prefix="test3",
    run_esmc=True,
    run_probert=True,
    cell_id_col="cell_id",
    sample_id_col="sample_id",
    reference=True,

)

pipe.print_concat_summary(results["esmc_concat"])
```
